```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data1 = np.load('results/honeybee.npz')
data2 = np.load('results/warm_honeybee.npz')

precision1 = data1['precision']
recall1 = data1['recall']
f11 = data1['f1']

precision2 = data2['precision']
recall2 = data2['recall']
f12 = data2['f1']

steps = range(len(precision1))

fig, axs = plt.subplots(1, 3, figsize=(20, 5))

axs[0].plot(steps, precision1, label='Honeybee Precision')
axs[0].plot(steps, precision2, label='Warm Honeybee Precision')
axs[0].set_title('Precision')
axs[0].legend()

axs[1].plot(steps, recall1, label='Honeybee Recall')
axs[1].plot(steps, recall2, label='Warm Honeybee Recall')
axs[1].set_title('Recall')
axs[1].legend()

axs[2].plot(steps, f11, label='Honeybee F1')
axs[2].plot(steps, f12, label='Warm Honeybee F1')
axs[2].set_title('F1')
axs[2].legend()

plt.tight_layout()
plt.show()
```
